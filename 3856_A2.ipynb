### Question 1

The multiclass log-loss is defined as
$$
L=-\frac{1}{n}\sum_{i=1}^{n}\sum_{k=1}^{K}I(y_i=k)\log\hat{p}_{i,k},
$$

where $I(y_i=k)$ is the indicator function, it equals $1$ if the true class of sample $i$ is $k$, and $0$ otherwise. $\hat{p}_{i,k}$ is the predicted probability that sample $i$ belongs to class $k$.

Since $I(y_i=k)=1$ only for the true class $y_i$, all other terms vanish. Therefore,

$$
L=-\frac{1}{n}\sum_{i=1}^{n}\log\hat{p}_{i,y_i}
$$

This shows that log-loss depends only on the predicted probabilities assigned to the true classes.

Furthermore, changing the numerical encoding of the class labels does not change the underlying classes or their corresponding predicted probabilities. For example, whether dog and cat are encoded as $\{0,1\}$ or $\{1,-1\}$, the loss remains unchanged as long as the predicted probabilities for the corresponding classes remain the same.

Hence, the multiclass log-loss is independent of the numerical encoding of the class labels.

### Question 2(a)

The sample Pearson correlation coefficient is 

$$
r=\frac{\sum_{i=1}^{n}(x_i-\bar{x})(y_i-\bar{y})}{\sqrt{\sum_{i=1}^{n}(x_i-\bar{x})^2}\sqrt{\sum_{i=1}^{n}(y_i-\bar{y})^2}}.
$$

Let $\mathbf{u}=\mathbf{x}-\bar{x}\mathbf{1}$ and $\mathbf{v}=\mathbf{y}-\bar{y}\mathbf{1}$ be the centered vectors.

By the definition of the dot product,

$$
\mathbf{u}^{\top}\mathbf{v}
=\sum_{i=1}^{n}(x_i-\bar{x})(y_i-\bar{y}).
$$

By  the definition of the Euclidean norm,

$$
\|\mathbf{u}\|
=\sqrt{\mathbf{u}^{\top}\mathbf{u}}
=\sqrt{\sum_{i=1}^{n}(x_i-\bar{x})^2},
$$

$$
\|\mathbf{v}\|
=\sqrt{\mathbf{v}^{\top}\mathbf{v}}
=\sqrt{\sum_{i=1}^{n}(y_i-\bar{y})^2}.
$$

Hence, the Pearson correlation can be expressed as

$$
r=\frac{\mathbf{u}^{\top}\mathbf{v}}{\|\mathbf{u}\|\|\mathbf{v}\|}.
$$

By the Cauchy-Schwarz inequality,

$$
|\mathbf{u}^{\top}\mathbf{v}|\leq\|\mathbf{u}\|\|\mathbf{v}\|.
$$

Assuming $s_x, s_y>0$, we have $\|\mathbf{u}\|\|\mathbf{v}\|>0$. Dividing both sides gives

$$
\frac{|\mathbf{u}^{\top}\mathbf{v}|}{\|\mathbf{u}\|\|\mathbf{v}\|}\leq1.
$$

Therefore,

$$
|r|\leq1\quad\Longrightarrow\quad -1\leq r\leq1.
$$

### Question 2(b)

We should prove both directions for if and only if questions.

**($\Rightarrow$) Suppose $r=1$.**

From part (a),

$$
r=\frac{\mathbf{u}^{\top}\mathbf{v}}{\|\mathbf{u}\|\|\mathbf{v}\|}=1.
$$

Hence,

$$
\mathbf{u}^{\top}\mathbf{v}=\|\mathbf{u}\|\|\mathbf{v}\|>0.
$$

Equality holds in the Cauchy-Schwarz inequality, so $\mathbf{v}=a\mathbf{u}$ for some $a\neq 0$.

Since

$$
\mathbf{u}^{\top}\mathbf{v}=a\|\mathbf{u}\|^2>0,
$$

we must have $a>0$.

Substituting the centered vectors gives

$$
\mathbf{y}-\bar{y}\mathbf{1}
=a(\mathbf{x}-\bar{x}\mathbf{1}).
$$

Rearranging,

$$
\mathbf{y}=a\mathbf{x}+(\bar{y}-a\bar{x})\mathbf{1}.
$$

Let $b=\bar{y}-a\bar{x}$. Therefore,

$$
y_i=ax_i+b,\quad a>0,\quad i=1,\ldots,n.
$$

**($\Leftarrow$) suppose $y_i=ax_i+b$ for all $i$, with $a>0$.**

Taking sample means gives

$$
\bar{y}=a\bar{x}+b.
$$

Thus,

$$
y_i-\bar{y}=a(x_i-\bar{x}),
$$

which implies $\mathbf{v}=a\mathbf{u}$.

Substituting into the Pearson correlation formula,

$$
\begin{aligned}
r
&=\frac{\mathbf{u}^{\top}(a\mathbf{u})}
{\|\mathbf{u}\|\|a\mathbf{u}\|}\\
&=\frac{a\|\mathbf{u}\|^2}{a\|\mathbf{u}\|^2}\\
&=1,
\end{aligned}
$$

since $a>0$.

Therefore,

$$
r=1\iff y_i=ax_i+b\text{ for all }i,\text{ with }a>0.
$$

### Question 2(c)

The proof is analogous to part(b), but with $a<0$.

**($\Rightarrow$)** If $r=-1$, then

$$
\mathbf{u}^{\top}\mathbf{v}
=-\|\mathbf{u}\|\|\mathbf{v}\|<0.
$$

By the equality condition of Cauchy-Schwarz, $\mathbf{v}=a\mathbf{u}$. Since

$$
a\|\mathbf{u}\|^2<0,
$$

we have $a<0$. As in part (b), this implies $y_i=ax_i+b$, where $b=\bar{y}-a\bar{x}$.

**($\Leftarrow$)** Conversely, if $y_i=ax_i+b$ with $a<0$, then $\mathbf{v}=a\mathbf{u}$. Thus,

$$
r=\frac{a\|\mathbf{u}\|^2}{|a|\|\mathbf{u}\|^2}
=\frac{a}{|a|}=-1.
$$

Therefore,

$$
r=-1\iff y_i=ax_i+b,\quad a<0.
$$



### Question 2(d)

Let the transformed observations be

$$
x_i'=\alpha x_i+\beta,\qquad
y_i'=\gamma y_i+\delta,

where $\alpha,\gamma\neq0$.

The transformed sample means are

$$

\bar{y}'=\gamma\bar{y}+\delta.

Therefore, the centered observations satisfy

$$
\begin{aligned}
x_i'-\bar{x}'

&=\alpha(x_i-\bar{x}),
\end{aligned}
$$

similary,

$$
y_i'-\bar{y}'=\gamma(y_i-\bar{y}).

Hence, the transformed centered vectors are

Using the Pearson correlation forula from part (a),

$$
\begin{aligned}
r'

### Question 3(a)

We prove by mathmatical induction that a neural network with identity activation functions can be expressed as

$$
f(x)=W_{\mathrm{eff}x+b_{\mathrm{eff}}}
$$

for any $L\geq0$.

**Base case ($L=0$):**

with no hidden layers, the network consists of only the output layer:

$$
f(x)=W^{(1)}x+b^{(1)}.
$$

Thus, $W_{\mathrm{eff}}=W^{{1}}$ and $b_{\mathrm{eff}}=b^{{1}}$. The base case holds.

**Inductive hypothesis:**

Assume that for $L=m$, the network can be written as

$$
f_m(x)=W_{\mathrm{eff}}^{{m}}x+b_{\mathrm{eff}}^{{m}}.
$$

**Inductive step:**

For $L=m+1$, adding one more layer gives

$$
f_{m+1}(x)=W^{(m+2)}f_m(x)+b^{(m+2)}.
$$

Substituting the inductive hypothesis,

$$
\begin{aligned}
f_{m+1}(x)
&=W^{(m+2)}(W_{\mathrm{eff}}^{(m)}x+b_{\mathrm{eff}}^{(m)})+b^{(m+2)}\\
&=W^{(m+2)}W_{\mathrm{eff}}^{(m)}x
+W^{(m+2)}b_{\mathrm{eff}}^{(m)}+b^{(m+2)}.
\end{aligned}
$$

Define

$$
W_{\mathrm{eff}}^{{m+1}}
=W^{(m+2)}W_{\mathrm{eff}}^{(m)}
$$

and

$$
b_{\mathrm{eff}}^{(m+1)}
=W^{(m+2)}b_{\mathrm{eff}}^{(m)}+b^{(m+2)}.
$$

Therefore,

$$
f_{m+1}(x)
=W_{\mathrm{eff}}^{(m+1)}x+b_{\mathrm{eff}}^{(m+1)}.
$$

Hence, by mathematical induction, the result holds for all $L\geq0$.

**General expressions:**

The effective weight matrix is

$$
W_{\mathrm{eff}}
=W^{(L+1)}W^{(L)}\cdots W^{(1)}.
$$

The effective bias vector is

$$
b_{\mathrm{eff}}
=b^{(L+1)}
+\sum_{j=1}^{L}
\left(W^{(L+1)}W^{(L)})\cdots W^{(j+1)}\right)b^{(j)}.
$$

Thus, a neural network with identity activation functions is equivalent to a single affine transformation.


### Question 3(b)

From part (a), the effective neural network model is

$$
f(x)=W_{\mathrm{eff}}x+b_{\mathrm{eff}}.
$$

The multiple linear regression model studied in Week 1 has the form

$$
\hat{y}=\beta_0+\sum_{j=1}^{p}\beta_jx_j
=\boldsymbol{\beta}^{\top}x+\beta_0.
$$

For a scalar output, identifying

$$
W_{\mathrm{eff}}=\boldsymbol{\beta}^{\top},
\qquad b_{\mathrm{eff}}=\beta_0,
$$

gives

$$
f(x)=\boldsymbol{\beta}^{\top}x+\beta_0=\hat{y}.
$$

Therefore, the effective neural network has the same functional form as a linear regression model. Multiple layers with identity activations do not introduce additional nonlinearity.



### Question 3(c)

From parts (a) and (b), a neural network using only identity activation functions can be reduced to a single affine transformation,

$$
f(x)=W_{\mathrm{eff}}x+b_{\mathrm{eff}},
$$

which has the same functional form as linear regression, regardless of the number of hidden layers.

Therefore, increasing the depth without nonlinear activation functions does not increase the network's ability to represent nonlinear relationships. Nonlinear activation functions, such as ReLU and sigmoid, allow the network to learn more complex nonlinear patterns that cannot be represented by a single affine transformation.

Overall, nonlinear activation functions are essential for increasing the expressive power of deep neural networks beyond linear models.

```python
# ME Task 2:
depths = [3,5,7]
results ={}

for depth in depths:
    accuracies = []

    for seed in seeds:
        X_train, X_test, y_train, y_test = train_test_split(
            X, y, test_size=0.2, random_state=seed
        )
        scaler = StandardScaler()
        X_train_scaled = scaler.fit_transform(X_train)
        X_test_scaled = scaler.transform(X_test)

        model = MLPClassifier(
            hidden_layer_sizes=(10,) * depth,
            activation="relu",
            solver="sgd",
            learning_rate_init=0.01,
            max_iter=100,
            batch_size=64,
            random_state=seed
        )
        model.fit(X_train_scaled, y_train)
        y_pred = model.predict(X_test_scaled)
        accuracy = accuracy_score(y_test, y_pred)

        accuracies.append(accuracy)

    results[depth] = np.mean(accuracies)

for depth, mean_accuracy in results.items():
    print(f"{depth} hidden layers: {mean_accuracy:.4f}")


        
```

```python
# ME Task 4:

optimizers = ["sgd", "adam"]
results_optimizer = {}

for optimizer in optimizers:
    accuracies = []

    for seed in seeds:
        X_train, X_test, y_train, y_test = train_test_split(
            X, y, test_size=0.2, random_state=seed
        )

        scaler = StandardScaler()
        X_train_scaled = scaler.fit_transform(X_train)
        X_test_scaled = scaler.transform(X_test)

        model = MLPClassifier(
            hidden_layer_sizes=(10,) * 7,
            activation="relu",
            solver=optimizer,
            learning_rate_init=0.01,
            max_iter=100,
            batch_size=64,
            random_state=seed
        )
        model.fit(X_train_scaled, y_train)
        y_pred = model.predict(X_test_scaled)
        accuracy = accuracy_score(y_test, y_pred)

        accuracies.append(accuracy)

    results_optimizer[optimizer] = np.mean(accuracies)

for optimizer, mean_accuracy in results_optimizer.items():
    print(f"{optimizer}: {mean_accuracy:.4f}")
```
